```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "K-Nearest Neighbors",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy": [
        0.958130,
        0.986523,
        0.998641,
        0.998622
    ],

    "Precision": [
        0.966703,
        0.986667,
        0.998641,
        0.998622
    ],

    "Recall": [
        0.958130,
        0.986523,
        0.998641,
        0.998622
    ],

    "F1-Score": [
        0.960103,
        0.986588,
        0.998641,
        0.998622
    ]
})

comparison
```

```python
comparison.to_csv("../results/model_comparison.csv", index=False)
```

```python
plt.figure(figsize=(4,5))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.title("Accuracy Comparison of Machine Learning Models")
plt.ylabel("Accuracy")
plt.ylim(0.94, 1.0)
plt.xticks(rotation=15)

plt.tight_layout()

plt.savefig("../images/model_accuracy_comparison.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(4,5))

plt.bar(
    comparison["Model"],
    comparison["F1-Score"]
)

plt.title("F1 Score Comparison")
plt.ylabel("F1 Score")
plt.ylim(0.94, 1.0)
plt.xticks(rotation=15)

plt.tight_layout()

plt.savefig("../images/model_f1_comparison.png", dpi=300)

plt.show()
```

```python
best_model = comparison.loc[
    comparison["Accuracy"].idxmax()
]

print("Best Model:\n")
print(best_model)
```

```python
comparison.sort_values(
    by="Accuracy",
    ascending=False
)
```

# Conclusion

Four supervised machine learning algorithms were evaluated for cyber attack detection using the CICIDS2017 dataset.

- Decision Tree achieved the highest accuracy (99.8641%).
- Random Forest achieved nearly identical performance (99.8622%) while generally providing better robustness.
- K-Nearest Neighbors (KNN) performed well but required significantly more prediction time due to its distance-based approach.
- Logistic Regression achieved the lowest accuracy because it assumes linear decision boundaries, whereas network traffic data contains more complex patterns.

Although Decision Tree achieved the highest accuracy, Random Forest was selected as the final model for the cyber attack detection system because it provides nearly identical predictive performance while generally reducing the risk of overfitting and improving robustness on unseen network traffic.
